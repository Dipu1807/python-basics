```python
#creating a dtataframe
students= pd.DataFrame({
    "name" :["aarav","aman","rohit","simran","neha","tania"],
    "branch":["cse","btech","cse","cse","cse","cse"],
    "marks": [85,90,np.nan,78,95,89],
    "hour_studies":[8,5,4,6,3,5],
})
students
```

```python
#check shape,columns,datatypes
print("shape :",students.shape)
print("columns :",list(students.columns))
print("dtypes :\n",students.dtypes)
print()
students.info()
```
